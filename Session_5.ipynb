```python
#Image Negative
import cv2
import matplotlib.pyplot as plt
import numpy as np

img = cv2.imread(r"Lena.png", 0)

negative_image = 255 - img

plt.figure(figsize=(10,5))


plt.subplot(1,2,1)
plt.imshow(img, cmap='gray')
plt.title('original')
plt.axis("off")


plt.subplot(1,2,2)
plt.imshow(negative_image, cmap='gray')
plt.title('nagetive Image')
plt.axis("off")

```

```python
#Logarithmic Transformation
import numpy as np
C = float(input("Give the Scaling Factor: "))

logTransformedImage = C * np.log1p(img)
logTransformedImage = np.clip(logTransformedImage,0,255).astype('uint8')

plt.figure(figsize=(10,5))


plt.subplot(1,2,1)
plt.imshow(img, cmap='gray')
plt.title('original')
plt.axis("off")


plt.subplot(1,2,2)
plt.imshow(logTransformedImage, cmap='gray')
plt.title('Log Transformed Image')
plt.axis("off")
```

```python
#Power Law Transformation ( s= c*r^gamma) (more than 1 -> img goes darker and Viceversa)
gamma = float(input("Enter the Gamma Value: "))

gamma_img =  (255*(img/255)**gamma).astype('uint8')

plt.figure(figsize=(10,5))


plt.subplot(1,2,1)
plt.imshow(img, cmap='gray')
plt.title('original')
plt.axis("off")


plt.subplot(1,2,2)
plt.imshow(gamma_img, cmap='gray')
plt.title('Gamma Transformed Image')
plt.axis("off")
```

```python
#Contrast Streching (automatic)
contrastDeprivedImage = cv2.imread(r"img4.tif")

contrastStrechedImage = cv2.normalize(contrastDeprivedImage, None, 0, 255, cv2.NORM_MINMAX)

plt.figure(figsize=(10,5))


plt.subplot(1,2,1)
plt.imshow(contrastDeprivedImage, cmap='gray')
plt.title('original')
plt.axis("off")


plt.subplot(1,2,2)
plt.imshow(contrastStrechedImage, cmap='gray')
plt.title('Contrast Streched Image')
plt.axis("off")
```

```python
#plotting Histogram of an image

plt.figure(figsize=(15,5))


plt.subplot(1,3,1)
plt.hist(img.ravel(), bins=256, range=[0,256])#Lena Image
plt.title('original(Lena)')


plt.subplot(1,3,2)
plt.hist(contrastDeprivedImage.ravel(),bins=256, range=[0,256])
plt.title('Contrast Deprived Image(Dark)')



plt.subplot(1,3,3)
plt.hist(contrastStrechedImage.ravel(),bins=256, range=[0,256])
plt.title('contrast Streched image(White)')

plt.show
```

```python
#Histogram Equalization
EqualizedImage = cv2.equalizeHist(img)

plt.figure(figsize=(10,5))


plt.subplot(1,2,1)
plt.imshow(img, cmap='gray')
plt.title('original')
plt.axis("off")


plt.subplot(1,2,2)
plt.imshow(EqualizedImage, cmap='gray')
plt.title('Histogram Equalized Image')
plt.axis("off")
```

```python
#Histogram Equalization using our own code

hist = np.zeros(256)

for row in img:
    for pixel in row:
        hist[pixel]+=1

cdf = hist.cumsum()

cdf_normalized = (cdf - cdf.min())*255/(cdf.max() - cdf.min())
cdf_normalized = cdf_normalized.astype("uint8")

Equalized_image_manual = cdf_normalized[img]

plt.figure(figsize=(10,10))


plt.subplot(2,2,1)
plt.imshow(img, cmap='gray')
plt.title('original')
plt.axis("off")

plt.subplot(2,2,2)
plt.hist(img.ravel(), bins=256, range=[0,256])#Lena Image
plt.title('original(Lena)')

plt.subplot(2,2,3)
plt.imshow(Equalized_image_manual, cmap='gray')
plt.title('Histogram Equalized Image')
plt.axis("off")

plt.subplot(2,2,4)
plt.hist(Equalized_image_manual.ravel(),bins=256,range=[0,256])
plt.title('Histogram Equalized Image Histogram')
```
